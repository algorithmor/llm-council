# ChildNet GUI: predict a child's face from both parents

A point-and-click interface for [ChildNet](https://github.com/MartinPernus/ChildNet) (Pernuš et al., IEEE Access 2023), with sliders for the **dominant parent**, the child's **age** and **gender**, and more.

1. *Runtime → Change runtime type* → choose a **GPU** (a T4 is fine). It also runs on CPU, just slower.
2. *Runtime → Run all*. Setup takes a few minutes, mostly downloading ChildNet's weights (3.6 GB unpacked).
3. Open the `https://….gradio.live` link that the last cell prints.

Stop the app by stopping the last cell. Anyone with the link can use the app while it runs.
